# if文の練習

問１：年齢を尋ねて、18歳以上かそうでないかに応じてメッセージを出力するPythonプログラムをif文を使って書いて下さい。

```python
age = int(input("年齢を入力してください: "))

if age >= 18:
    print("あなたは18歳以上です。")
else:
    print("あなたは18歳未満です。")
```

問２：問１で作成したプログラムについて、下記の点を確認して下さい。

* 全角数字で年齢を入力したら、どうなりますか？
* `15歳` のように入力したら、どうなりますか？
* `24です` のように入力したら、どうなりますか？
* `-3` のように負の数を入力したら、どうなりますか？
* `10018` のような大きな数を入力したら、どうなりますか？
* `3.6` のように小数を入力したら、どうなりますか？

**実用システム**を開発するときは、このような、**ありとあらゆる入力に対して、適切に応答する必要**があります。考えられるすべてのケースに対して適切に応答できるよう、プログラムを改良して下さい。



```python
import unicodedata

def get_valid_age():
    while True:
        age_str = input("年齢を入力してください: ")

        # 全角数字を半角に変換し、余分な空白を除去
        age_str_normalized = unicodedata.normalize('NFKC', age_str).strip()

        try:
            age = int(age_str_normalized)

            if age < 0:
                print("年齢に負の数は入力できません。")
            elif age > 120: # 一般的な人間の寿命を考慮
                print(f"年齢 {age} は現実的ではありません。もう一度入力してください。")
            else:
                # 有効な年齢が得られたらループを抜ける
                return age
        except ValueError:
            # int()変換でエラーが出た場合 (例: '15歳', '24です', '3.6', 'abc')
            print("無効な入力です。数字のみを入力してください。")

# 年齢の入力と検証
age = get_valid_age()

# 18歳以上かどうかの判定
if age >= 18:
    print("あなたは18歳以上です。")
else:
    print("あなたは18歳未満です。")
```

# for文の練習

問３：整数のリストから、3の倍数だけを取り出すPythonプログラムをfor文を使って書いて下さい。

（参考）リスト内包表記を使うと、より簡潔に書くことができます。

3の倍数と3がつく数字のときにアホになる面白いプログラムを作って

```python
for i in range(1, 41):
    if i % 3 == 0 or '3' in str(i):
        print(f"{i} 🤪ぁほぉ！")
    else:
        print(i)
```

```python
from gtts import gTTS
from IPython.display import Audio, display

# アホな読み方を返す関数
def get_aho_reading(n):
    aho_dict = {
        3: "さぁーーん", 6: "ろぉーーく", 9: "きゅぅーーう",
        12: "じゅうにぃーー", 13: "じゅうさぁーーん", 15: "じゅうごぉーー",
        18: "じゅうはちぃーー", 21: "にじゅういぃーーち", 23: "にじゅうさぁーーん",
        24: "にじゅうよぉーーん", 27: "にじゅうななぁーー", 30: "さぁーーんじゅぅーー"
    }
    if n in aho_dict:
        return aho_dict[n]
    if 31 <= n <= 39:
        base = "さぁーーんじゅう"
        sub_dict = {1: "いぃーーち", 2: "にぃーー", 3: "さぁーーん", 4: "よぉーーん", 5: "ごぉーー", 6: "ろぉーーく", 7: "ななぁーー", 8: "はちぃーー", 9: "きゅぅーーう"}
        return base + sub_dict.get(n - 30, "")
    return str(n)

# 読み上げるテキストを結合するための変数
spoken_text = ""

for i in range(1, 41):
    if i % 3 == 0 or '3' in str(i):
        # アホな読み方に変換
        spoken_text += f"{get_aho_reading(i)}、 "
    else:
        spoken_text += f"{i}。 "

print("音声を生成中...")

# gTTSで日本語の音声を生成
tts = gTTS(text=spoken_text, lang='ja')
tts.save('nabeatsu_aho.mp3')

print("生成完了！再生します。")

# Colab上で音声を再生
display(Audio('nabeatsu_aho.mp3', autoplay=True))
```

アホになるときに読み上げスピードをゆっくりにして

```python
from gtts import gTTS
from IPython.display import Audio, display
import os

# アホな読み方を返す関数
def get_aho_reading(n):
    aho_dict = {
        3: "さぁーーん", 6: "ろぉーーく", 9: "きゅぅーーう",
        12: "じゅうにぃーー", 13: "じゅうさぁーーん", 15: "じゅうごぉーー",
        18: "じゅうはちぃーー", 21: "にじゅういぃーーち", 23: "にじゅうさぁーーん",
        24: "にじゅうよぉーーん", 27: "にじゅうななぁーー", 30: "さぁーーんじゅぅーー"
    }
    if n in aho_dict:
        return aho_dict[n]
    if 31 <= n <= 39:
        base = "さぁーーんじゅう"
        sub_dict = {1: "いぃーーち", 2: "にぃーー", 3: "さぁーーん", 4: "よぉーーん", 5: "ごぉーー", 6: "ろぉーーく", 7: "ななぁーー", 8: "はちぃーー", 9: "きゅぅーーう"}
        return base + sub_dict.get(n - 30, "")
    return str(n)

print("音声を生成中... (少し時間がかかります)")

# 複数のMP3を結合して1つのファイルにする
with open("nabeatsu_mixed.mp3", "wb") as outfile:
    for i in range(1, 41):
        if i % 3 == 0 or '3' in str(i):
            text = f"{get_aho_reading(i)}、"
            # アホになるときはゆっくり読む
            tts = gTTS(text=text, lang='ja', slow=True)
        else:
            text = f"{i}。"
            # 普通の数字は通常の速さで読む
            tts = gTTS(text=text, lang='ja', slow=False)

        # 一時ファイルとして保存
        temp_filename = f"temp_{i}.mp3"
        tts.save(temp_filename)

        # 一時ファイルの中身を結合用ファイルに追記
        with open(temp_filename, "rb") as infile:
            outfile.write(infile.read())

        # 一時ファイルを削除
        os.remove(temp_filename)

print("生成完了！再生します。")

# Colab上で音声を再生
display(Audio('nabeatsu_mixed.mp3', autoplay=True))
```
