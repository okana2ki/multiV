# 単回帰分析の決定係数が1になるようなデータの例

**ユーザ：**

単回帰分析の決定係数が1になるようなデータの例を示す。

**Gemini 3.1 Pro（有料）：**

単回帰分析の決定係数（$R^2$）が1になるのは、すべてのデータポイントが完全に一直線上に並んでいる（誤差がゼロの）場合です。

既存のセルに、そのようなデータ（例: $y = 2x + 1$）を作成して回帰分析を行い、決定係数が1になることを確認してグラフに描画するコードを追加しました。

コードが正常に実行されました。

出力結果からわかるように、決定係数 ($R^2$) は 1.0 と計算されました。また、プロットされたグラフを見ると、すべての青いデータポイントが赤い回帰直線上に完全に重なっていることが確認できます。このように誤差が全くないデータセットにおいて、決定係数は最大の1となります。

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 決定係数が1になるように、完全に直線（y = 2x + 1）上に乗るデータを作成
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([3, 5, 7, 9, 11])

# 単回帰モデルの作成と学習
model = LinearRegression()
model.fit(X, y)

# 決定係数 (R^2) の計算
r2_score = model.score(X, y)
print(f"決定係数 (R^2): {r2_score}")

# データのプロット
plt.scatter(X, y, color='blue', label='Data points')
plt.plot(X, model.predict(X), color='red', linestyle='--', label='Regression line')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Perfect Linear Regression (R^2 = 1.0)')
plt.legend()
plt.grid(True)
plt.show()
```

# 単回帰分析の決定係数が0になるようなデータの例

**ユーザ：**

単回帰分析の決定係数が0になるようなデータの例を示す。

**Gemini 3.1 Pro:**

単回帰分析の決定係数（$R^2$$R^2$）が0になるのは、Xとyの間に線形な関係が全くない場合です。たとえば、データがU字型に分布しているなど、回帰直線が水平（傾き0）になり、yのばらつきを全く説明できない場合に決定係数は0になります。

そのようなデータを作成し、回帰分析を行ってグラフに描画するコードを追加しました。

コードが正常に実行されました。

出力結果にある 決定係数 (R^2): -2.220446049250313e-16 は、コンピュータの計算上のわずかな誤差（浮動小数点誤差）によるもので、実質的に 0 です。

プロットされたグラフを見ると、データがU字型に分布しているため、線形回帰モデルはこれらのデータの傾向を直線で捉えることができず、回帰直線が完全に水平になっています。これは、Xの値からyの値を予測するのに役立つ線形な関係が全くないことを示しており、その結果として決定係数が0になります。

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 決定係数が0になるように、線形関係のないデータを作成（例：U字型の左右対称なデータ）
X_zero = np.array([[1], [2], [3], [4], [5]])
y_zero = np.array([2, 1, 0, 1, 2])

# 単回帰モデルの作成と学習
model_zero = LinearRegression()
model_zero.fit(X_zero, y_zero)

# 決定係数 (R^2) の計算
r2_score_zero = model_zero.score(X_zero, y_zero)
print(f"決定係数 (R^2): {r2_score_zero}")

# データのプロット
plt.scatter(X_zero, y_zero, color='green', label='Data points')
plt.plot(X_zero, model_zero.predict(X_zero), color='red', linestyle='--', label='Regression line')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Zero Linear Correlation (R^2 = 0.0)')
plt.legend()
plt.grid(True)
plt.show()
```

**ユーザ：**

非常に小さい値であるとはいえ、決定係数が正確に0ではない理由を教えて

**Gemini 3.1 Pro:**

決定係数が正確に 0.0 ではなく -2.220446049250313e-16 という非常にゼロに近い値になった理由は、**コンピュータの「浮動小数点演算」に伴う微小な計算誤差（丸め誤差）** によるものです。

コンピュータは内部で数値を2進数（0と1）で扱っていますが、一部の小数を2進数で完全に正確に表現することができず、ごくわずかなズレが生じます。

scikit-learnの内部で決定係数（$R^2$）を計算する際、分散や誤差の二乗和などの割り算や引き算が行われます。理論上は「$1 - 1 = 0$」になるはずの計算が、この微小なズレの蓄積によって「$1 - 1.0000000000000002 = -0.0000000000000002$」のように計算されてしまうためです。

この -2.22...e-16 という値は、データサイエンスやプログラミングにおいては実質的に「完全にゼロ」として扱って全く問題ありません。

**ユーザ：**

4個のデータが正方形の4つの頂点上に位置しているときや、ひし形の4つの頂点上に位置しているときも、決定係数はゼロになりますか？

**Gemini 3.1 Pro:**

非常に鋭い質問ですね！結論から言うと、正方形の場合はどのような向きであっても常に決定係数はゼロになりますが、**ひし形の場合は「向き（傾き）」によって決定係数がゼロにならない（0より大きくなる）**ことがあります。

理由を視覚的に理解できるよう、以下の4つのパターン（軸に平行な正方形、傾いた正方形、軸に平行なひし形、傾いたひし形）で回帰分析を行い、決定係数を確認するコードを作成しました。実行して結果を見てみましょう。

コードが正常に実行され、グラフが描画されました。

グラフからわかるように：

1. 正方形（上段2つ）：軸に平行な場合も、45度傾けた場合も、決定係数（$R^2$）は 0.0 となりました。
2. ひし形（下段2つ）：軸に平行な場合は対称性により $R^2$ は 0.0 ですが、45度傾けた場合はデータが斜めに分布するため相関が生じ、$R^2$ が 0.3600 とゼロより大きくなりました。

このように、正方形の場合は向きに関わらず決定係数はゼロになりますが、ひし形の場合は傾き具合によって決定係数が変化します。

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

def calc_r2_and_plot(ax, X, y, title):
    model = LinearRegression()
    X_reshaped = X.reshape(-1, 1)
    model.fit(X_reshaped, y)
    r2 = model.score(X_reshaped, y)

    # 散布図のプロット
    ax.scatter(X, y, color='purple', s=100, zorder=5)

    # 図形の輪郭を描画（点を順番に結ぶ）
    ax.plot(np.append(X, X[0]), np.append(y, y[0]), color='gray', linestyle='dotted')

    # 回帰直線を描画
    x_line = np.linspace(min(X)-1, max(X)+1, 10)
    y_line = model.predict(x_line.reshape(-1, 1))
    ax.plot(x_line, y_line, color='red', linestyle='-')

    # 数値誤差で極小のマイナスになる場合は0とする
    if abs(r2) < 1e-15:
        r2 = 0.0

    ax.set_title(f'{title}\n$R^2$ = {r2:.4f}')
    ax.set_aspect('equal')
    ax.grid(True)

fig, axes = plt.subplots(2, 2, figsize=(10, 10))

# 1. 軸に平行な正方形
X_sq1 = np.array([-1, 1, 1, -1])
y_sq1 = np.array([1, 1, -1, -1])
calc_r2_and_plot(axes[0, 0], X_sq1, y_sq1, "Axis-aligned Square")

# 2. 傾いた正方形 (45度)
X_sq2 = np.array([0, 1, 0, -1])
y_sq2 = np.array([1, 0, -1, 0])
calc_r2_and_plot(axes[0, 1], X_sq2, y_sq2, "Rotated Square")

# 3. 軸に平行なひし形
X_rh1 = np.array([0, 2, 0, -2])
y_rh1 = np.array([1, 0, -1, 0])
calc_r2_and_plot(axes[1, 0], X_rh1, y_rh1, "Axis-aligned Rhombus")

# 4. 傾いたひし形 (45度回転)
angle = np.pi / 4
X_rh2 = X_rh1 * np.cos(angle) - y_rh1 * np.sin(angle)
y_rh2 = X_rh1 * np.sin(angle) + y_rh1 * np.cos(angle)
calc_r2_and_plot(axes[1, 1], X_rh2, y_rh2, "Rotated Rhombus")

plt.tight_layout()
plt.show()
```

**解説：**
* **正方形**は非常に高い対称性を持っているため、どれだけ回転させてもXとYの間に「全体的な右上がり・右下がり」という相関関係（共分散）が生まれず、常に $R^2 = 0$ になります。
* **ひし形**（正方形でないもの）の場合、対角線がX軸・Y軸と平行な状態であれば $R^2 = 0$ になりますが、図形が斜めに傾くと、データ全体が斜め方向に引き伸ばされたような分布になり、そこに線形な傾向（相関）が生まれるため、決定係数が0よりも大きくなります。

**ユーザ：**

非常に多数の点がランダムに配置されているときも、決定係数はゼロになりますか？

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 乱数のシードを固定（毎回同じ結果にするため）
np.random.seed(42)

# 10,000個のランダムなデータを生成（Xとyを完全に独立して生成）
N = 10000
X_rand = np.random.rand(N, 1)  # 0から1の範囲の乱数
y_rand = np.random.rand(N)     # 0から1の範囲の乱数

# 単回帰モデルの作成と学習
model_rand = LinearRegression()
model_rand.fit(X_rand, y_rand)

# 決定係数 (R^2) の計算
r2_score_rand = model_rand.score(X_rand, y_rand)

# データのプロット
plt.figure(figsize=(8, 6))
# 点が多いため、半透明（alpha=0.1）でプロット
plt.scatter(X_rand, y_rand, color='gray', alpha=0.1, s=10, label='Random points')

# 回帰直線のプロット
x_line = np.array([[0], [1]])
y_line = model_rand.predict(x_line)
plt.plot(x_line, y_line, color='red', linewidth=3, label='Regression line')

plt.xlabel('X')
plt.ylabel('y')
plt.title(f'10,000 Random Points\n$R^2$ = {r2_score_rand:.6f}')
plt.legend()
plt.grid(True)
plt.show()

print(f"決定係数 (R^2): {r2_score_rand}")
```

**解説：**

結果を見ると、$R^2$ は `0.0000...` のような非常にゼロに近い値になります。

* **完全にランダム**ということは、Xがどんな値であっても、yの値の予測には全く役に立たない（傾きがほぼ0の水平な直線になる）ことを意味します。
* データ数が少ない場合は、偶然による偏りで決定係数が少し大きくなることもありますが、データ数（点）が多ければ多いほど、その偶然の偏りが打ち消し合い、決定係数は理論値である **0** に限りなく収束していきます。

**ユーザ：**

決定係数が0になる他の面白い例はありませんか？

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 1. 円（Circle）
theta = np.linspace(0, 2*np.pi, 100)
X_circle = np.cos(theta).reshape(-1, 1)
y_circle = np.sin(theta)

model_circle = LinearRegression()
model_circle.fit(X_circle, y_circle)
r2_circle = model_circle.score(X_circle, y_circle)

axes[0].scatter(X_circle, y_circle, color='blue', label='Data')
axes[0].plot(X_circle, model_circle.predict(X_circle), color='red', linewidth=2, label='Regression line')
axes[0].set_title(f'Circle\n$R^2$ = {max(0.0, r2_circle):.4f}') # 誤差によるマイナスを0に丸める
axes[0].set_aspect('equal')
axes[0].grid(True)
axes[0].legend()

# 2. サイン波（Sine Wave）
X_sine = np.linspace(0, 4*np.pi, 200).reshape(-1, 1)
y_sine = np.sin(X_sine.flatten())

model_sine = LinearRegression()
model_sine.fit(X_sine, y_sine)
r2_sine = model_sine.score(X_sine, y_sine)

axes[1].scatter(X_sine, y_sine, color='green', label='Data')
axes[1].plot(X_sine, model_sine.predict(X_sine), color='red', linewidth=2, label='Regression line')
axes[1].set_title(f'Sine Wave\n$R^2$ = {max(0.0, r2_sine):.4f}')
axes[1].grid(True)
axes[1].legend()

plt.tight_layout()
plt.show()
```

**解説：**

* **円形（Circle）**：データは「$x^2 + y^2 = 1$」という完璧な数式（ルール）に従って配置されていますが、全体として右上がりでも右下がりでもないため、決定係数は0になります。
* **サイン波（Sine Wave）**：こちらも「$y = \sin(x)$」という完璧な法則がありますが、周期的に上がったり下がったりを繰り返すため、1つの直線で表現しようとすると水平な直線（傾き0）になってしまい、決定係数は0になります。

このように、**「$R^2 = 0$ だからといって、データ間に何の関係もないとは限らない」**というのがデータ分析における重要な教訓です。

**ユーザ：**

サイン波は、単回帰分析すると傾きがある直線になるので決定係数はゼロにはなりませんね。単回帰分析で傾きを求めて、傾きがゼロになるように、回転操作を加えると決定係数をゼロにできることを確認して。

**解説：**

ご指摘の通り、先ほどのサイン波（0から$4\pi$の範囲）には全体としてわずかな傾きが存在していたため、$R^2$ が約0.15となっていました。データの傾き（トレンド）を求め、それを打ち消すように逆回転させることで、傾きを完全に0にし、決定係数も0にすることができます。

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 元のサイン波データ (0 から 4π)
X_sine = np.linspace(0, 4*np.pi, 200).reshape(-1, 1)
y_sine = np.sin(X_sine.flatten())

# 1. 元のデータで単回帰分析を行い、傾き(slope)を求める
model_orig = LinearRegression()
model_orig.fit(X_sine, y_sine)
slope = model_orig.coef_[0]

# 2. 傾きから回転する角度（ラジアン）を計算
# 傾き m = tan(θ) なので、θ = arctan(m)
# 傾きを0にするには、-θ だけ回転させる
theta = np.arctan(slope)
angle = -theta

# 回転行列を使ってデータを回転させる
x_orig = X_sine.flatten()
# (x, y) 座標の回転公式
x_rot = x_orig * np.cos(angle) - y_sine * np.sin(angle)
y_rot = x_orig * np.sin(angle) + y_sine * np.cos(angle)

X_rot = x_rot.reshape(-1, 1)

# 3. 回転後のデータで再度単回帰分析を行う
model_rot = LinearRegression()
model_rot.fit(X_rot, y_rot)
r2_rot = model_rot.score(X_rot, y_rot)

# プロット
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 元のサイン波
axes[0].scatter(X_sine, y_sine, color='green', label='Original Data')
axes[0].plot(X_sine, model_orig.predict(X_sine), color='red', linewidth=2, label=f'Line (slope={slope:.4f})')
axes[0].set_title(f'Original Sine Wave\n$R^2$ = {model_orig.score(X_sine, y_sine):.4f}')
axes[0].grid(True)
axes[0].legend()

# 回転後のサイン波
axes[1].scatter(X_rot, y_rot, color='purple', label='Rotated Data')
axes[1].plot(X_rot, model_rot.predict(X_rot), color='red', linewidth=2, label=f'Line (slope={model_rot.coef_[0]:.4f})')
# マイナスの微小な誤差を0に丸める
r2_rot_display = max(0.0, r2_rot) if abs(r2_rot) < 1e-10 else r2_rot
axes[1].set_title(f'Rotated Sine Wave (Slope 0)\n$R^2$ = {r2_rot_display:.4f}')
axes[1].grid(True)
axes[1].legend()

plt.tight_layout()
plt.show()
```

**ユーザ：**

回転変換というのは、紙にプロットしてあるとすると、その紙を回転させて見ることと同じだと思っていたのですが、結果の散布図を見比べるとそうはなっていないように見えます。私は何か勘違いしているでしょうか？

**解説：**

おっしゃる通り、回転変換は「紙を回転させる」のと同じ操作です。前回のグラフでそう見えなかったのは、グラフのX軸とY軸のスケール（縮尺）が異なっていたためです。

以下のコードでは、`set_aspect('equal')` を用いてX軸とY軸の1メモリの長さを完全に同一にして描画しています。これにより、本来の平坦なサイン波が、形を変えずにそのまま（原点を中心に）少しだけ回転して水平になった様子が確認できます。

```python
import matplotlib.pyplot as plt

# 縦横比を1:1にしたグラフの描画
fig, axes = plt.subplots(2, 1, figsize=(10, 6))

# 元のサイン波 (スケール1:1)
axes[0].scatter(X_sine, y_sine, color='green', label='Original Data')
axes[0].plot(X_sine, model_orig.predict(X_sine), color='red', linewidth=1.5, label=f'Line (slope={slope:.4f})')
axes[0].set_title('Original Sine Wave (Aspect = Equal)')
axes[0].set_aspect('equal', adjustable='datalim')
axes[0].grid(True)
axes[0].legend(loc='upper right')

# 回転後のサイン波 (スケール1:1)
axes[1].scatter(X_rot, y_rot, color='purple', label='Rotated Data')
axes[1].plot(X_rot, model_rot.predict(X_rot), color='red', linewidth=1.5, label=f'Line (slope={model_rot.coef_[0]:.4f})')
axes[1].set_title('Rotated Sine Wave (Aspect = Equal)')
axes[1].set_aspect('equal', adjustable='datalim')
axes[1].grid(True)
axes[1].legend(loc='upper right')

plt.tight_layout()
plt.show()
```
